ganzer Code in einem Feld, sodass ich nur einmal ausführen muss:

```python
base_url = "https://quotes.toscrape.com/page/"

corpus = []

for page in range(1, 20):
    url = base_url + str(page)

    print("Requesting", url)
    response = get(url, headers={"User-Agent": user_agent})



    soup = BeautifulSoup(
        response.content,
        "html.parser"
    )


    quotes = soup.find_all(
        "div", 
        attrs={"class": "quote"}
    )

    for quote in quotes:

        text = quote.find_next("span", attrs={"class": "text"})
        corpus.append(text.text)


    sleep(5)
```

```python
corpus[10]
```

```python
for index, quote in enumerate(corpus):

    with open(f"output/quote_{index}.txt", "w", encoding="utf-8") as file:
        file.write(quote)

```
